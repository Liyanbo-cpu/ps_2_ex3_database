# Insight ② — The three-point revolution

**Question.** How did the three-point shot change from the first seasons with recorded attempts to 2022–23?

The database contains team box scores in `game`. Each row is one game, so the query combines home and away attempts. It reports average three-point attempts per team-game and the proportion of all field-goal attempts that came from three. The available values begin in 1981–82.

**Finding.** From 1981–82 to 2022–23, average three-point attempts rose from about 2.3 to 34.2 per team per game (about 15×). Threes grew from 2.6% to 38.7% of all field-goal attempts. In other words, nearly four of every ten shots in the latest recorded season were threes. This describes a change in shot selection; it does not by itself show that three-pointers caused scoring to rise.


## SQL

The full reusable query is in `queries/insight_2.sql`. It keeps regular-season IDs (`2` prefix), groups by season, and orders chronologically. `season_start_year` is extracted from the NBA season ID; for example, 1981 denotes 1981–82. The shot-share calculation uses total attempts as the denominator, rather than averaging game-level percentages.


```python
from pathlib import Path
from IPython.display import display
from nba.data_loader import connect_database
from nba.insight_2 import three_point_summary, three_point_trend
from nba.insight_2_viz import plot_three_point_trend

connection = connect_database(Path('data/nba.sqlite'))
trend = three_point_trend(connection)
summary = three_point_summary(trend)
connection.close()

display(trend.head())
display(trend.tail())
display(summary)
```

```python
figure = plot_three_point_trend(trend)
display(figure)
```

## Interpretation

The two lines answer slightly different questions: attempts per team-game show volume, while three-point share controls for the fact that teams also take different numbers of shots overall. Both rise steeply across the observed period. The query’s endpoint is the latest season in this database (2022–23), so it should not be described as the current NBA season.
